```python
import pandas as pd

# Cargar el archivo directamente desde tu entorno
df = pd.read_csv("muestras_calidad_agua.csv")

# Ver las primeras filas
df.head()
```

```python
# Extraer columnas correctas
df_oxigeno = df[['sitio', 'oxigeno']].copy()

# Clasificar nivel de oxígeno
def clasificar_nivel(o2):
    if o2 < 5:
        return "bajo"
    elif 5 <= o2 < 7:
        return "medio"
    else:
        return "alto"

df_oxigeno['nivel_oxigeno'] = df_oxigeno['oxigeno'].apply(clasificar_nivel)

# Crear columna de riesgo biológico
df_oxigeno['riesgo_biologico'] = df_oxigeno['oxigeno'].apply(lambda x: "Advertencia" if x < 5 else "✔️")

# Exportar resultados
df_oxigeno.to_csv("resultado_oxigeno.csv", index=False)

from google.colab import files
files.download("resultado_oxigeno.csv")

# Calcular porcentaje de sitios con riesgo
riesgo = df_oxigeno['riesgo_biologico'].value_counts()
porcentaje_riesgo = (riesgo.get("Advertencia", 0) / len(df_oxigeno)) * 100
print(f"Porcentaje de sitios con riesgo: {porcentaje_riesgo:.2f}%")
```
