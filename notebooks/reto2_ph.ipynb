# Reto 2 - Evaluación de pH y alerta química

## Objetivo
Analizar los valores de pH de diferentes cuerpos de agua en Cundinamarca utilizando condicionales en Python para identificar posibles riesgos químicos y clasificar la calidad del agua.

### Variables analizadas
- pH
- Clasificación química
- Nivel de alerta ambiental

### Autor
Estudiante 2


```python
import pandas as pd

print("Pandas cargado correctamente")
```

```python
df = pd.read_csv("../data/muestras_calidad_agua.csv")
```

```python
df 
```

```python
print("Promedio de pH:", df["pH"].mean())
print("pH mínimo:", df["pH"].min())
print("pH máximo:", df["pH"].max())
```

```python
def clasificar_ph(ph):

    if ph < 6.5:
        return "Ácida"

    elif ph <= 8.5:
        return "Neutra"

    else:
        return "Alcalina"
```

```python
df["clasificacion_pH"] = df["pH"].apply(clasificar_ph)
```

```python
df
```

```python
def alerta_quimica(ph):

    if ph < 6:
        return "⚠️ Alta acidez"

    elif ph < 6.5:
        return "⚠️ Riesgo moderado"

    elif ph <= 8.5:
        return "✅ Condición estable"

    else:
        return "⚠️ Riesgo por alcalinidad"
```

```python
df["alerta_quimica"] = df["pH"].apply(alerta_quimica)
```

```python
df
```

```python
sitios_riesgo = df[
    (df["pH"] < 6.5) | (df["pH"] > 8.5)
]

sitios_riesgo
```

```python
print("Número de sitios con posible alteración química:")
print(len(sitios_riesgo))
```
