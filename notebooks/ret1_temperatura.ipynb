```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/muestras_calidad_agua.csv')
df
```

```python
df_temp = df[['sitio', 'temperatura']]
df_temp
```

```python
df_temp['categoria_temp'] = np.where(
    df_temp['temperatura'] < 20,
    'fría',
    np.where(
        df_temp['temperatura'] <= 25,
        'templada',
        'caliente'
    )
)

df_temp
```

```python
df_temp['riesgo_termico'] = np.where(
    df_temp['temperatura'] > 25,
    'Sí',
    'No'
)

df_temp
```

```python
df_caliente = df_temp[df_temp['temperatura'] > 25]

df_caliente
```

```python
df_temp.to_csv('resultado_temperatura.csv', index=False)
```

```python
porcentaje = (
    (df_temp['riesgo_termico'] == 'Sí').sum()
    / len(df_temp)
) * 100

print(f'Porcentaje con riesgo térmico: {porcentaje:.1f}%')
```

En este reto se analiza la temperatura del agua en diferentes sitios de Cundinamarca utilizando condicionales y herraminetas pandas, para clasificar la temperatura de cada sitio y asi establecer si hay riesgo en el cuerpo de agua.
