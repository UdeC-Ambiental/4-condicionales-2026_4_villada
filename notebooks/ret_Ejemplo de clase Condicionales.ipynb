```python
temperatura = 28

if temperatura > 30:
    print("Día caluroso")

elif temperatura >= 20:
    print("Día templado")

else:
    print("Día frío")
```

```python
# Lista de valores de pH

ph_lista = [6.1, 7.4, 8.9]

# Seleccionamos un valor
ph = ph_lista[1]

print("Valor de pH:", ph)

if ph < 6.5:
    print("El agua es ácida")

elif ph <= 8.5:
    print("El agua es neutra")

else:
    print("El agua es básica")
```

```python
# Tupla: (temperatura, oxígeno)

muestra = (29, 4.2)

temperatura = muestra[0]
oxigeno = muestra[1]

print("Temperatura:", temperatura)
print("Oxígeno:", oxigeno)

if temperatura > 30 and oxigeno < 5:
    print("Condiciones críticas")

elif temperatura > 30:
    print("Alta temperatura")

elif oxigeno < 5:
    print("Bajo oxígeno")

else:
    print("Condiciones adecuadas")
```

```python
# Diccionario de una muestra

muestra = {
    "sitio": "Río Sur",
    "pH": 6.3,
    "oxigeno": 5.8
}

print(muestra)

if muestra["pH"] < 6.5:
    print("Agua ácida")

elif muestra["pH"] <= 8.5:
    print("Agua neutra")

else:
    print("Agua básica")


if muestra["oxigeno"] < 5:
    print("Oxígeno bajo")

else:
    print("Oxígeno adecuado")
```

```python
import pandas as pd

# Crear DataFrame

muestra = pd.DataFrame({
    "sitio": ["Quebrada Verde"],
    "pH": [8.7],
    "oxigeno": [4.5]
})

muestra
```

```python
valor_ph = muestra.loc[0, "pH"]

if valor_ph < 6.5:
    clasificacion = "ácida"

elif valor_ph <= 8.5:
    clasificacion = "neutra"

else:
    clasificacion = "básica"

print("Clasificación:", clasificacion)
```

```python
import numpy as np
import pandas as pd

datos = pd.DataFrame({
    "sitio": ["A", "B", "C"],
    "pH": [6.2, 7.5, 9.1]
})

datos
```

```python
datos["clasificacion"] = np.where(
    datos["pH"] < 6.5,
    "ácido",
    np.where(
        datos["pH"] <= 8.5,
        "neutro",
        "básico"
    )
)

datos
```
